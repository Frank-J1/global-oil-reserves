```python
import pandas as pd
```

```python
#We only care about the 3 page as of right now, consumption will be considered later
df_global_oil_withdraws = pd.read_excel(
    'data/STEO_m.xlsx',
    sheet_name='3atab',
    skiprows=34,
    nrows=4
)
```

```python
df_global_oil_withdraws.head()
```

```python
df_global_oil_withdraws = df_global_oil_withdraws.transpose().reset_index()

df_global_oil_withdraws.columns = [
    "World Total",
    "USA",
    "OECD",
    "Other Inventory draws and balance"
]

df_global_oil_withdraws = df_global_oil_withdraws.iloc[2:].reset_index(drop=True)
```

```python
#Add dates, MS = Month Start
df_global_oil_withdraws["Date"] = pd.date_range(
    start="2022-01-01",
    periods=len(df_global_oil_withdraws),
    freq="MS"
)
```

```python
#Move date to the start to reflect original excel sheet
col = ["Date"] + [col for col in df_global_oil_withdraws.columns if col != "Date"]
df_global_oil_withdraws = df_global_oil_withdraws[col]
```

```python
df_global_oil_withdraws = df_global_oil_withdraws[df_global_oil_withdraws['Date'] <= '2026-09-30']
```
