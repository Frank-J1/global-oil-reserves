```python
import pandas as pd
```

```python
df_global_oil_inventories = pd.read_excel(
    'data/STEO_m.xlsx',
    sheet_name='3atab',
    skiprows=40,
    nrows=2
)
```

```python
df_global_oil_inventories.head()
```

```python
df_global_oil_inventories = df_global_oil_inventories.transpose().reset_index()
```

```python
#VS code only displays so much but it did transpose properly
df_global_oil_inventories.head()
```

```python
df_global_oil_inventories.columns = df_global_oil_inventories.iloc[1]
df_global_oil_inventories = df_global_oil_inventories.iloc[2:].reset_index(drop=True)
```

```python
df_global_oil_inventories.columns.name = None
```

```python
#Add date col
df_global_oil_inventories["Date"] = pd.date_range(
    start="2022-01-01",
    periods=len(df_global_oil_inventories),
    freq="MS"
)
```

```python
#Move date col to the start
col = ["Date"] + [col for col in df_global_oil_inventories if col != "Date"]
df_global_oil_inventories = df_global_oil_inventories[col]
```

```python
df_global_oil_inventories = df_global_oil_inventories[df_global_oil_inventories['Date'] <= '2026-09-30']
```
